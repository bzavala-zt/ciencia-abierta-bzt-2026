# Creando una función 

```python
# conozco la longitud de los datos de temperatura
L = len(datos["Temperatura"])
```

```python
def promedio_variables(variable, nu_dato):
    """
    Se obtiene el promedio de las variables
    """
    suma = variable.sum()
    mm = suma/nu_dato

    return mm
```

```python
#Utilizamos la función establecida anteriormente
promedio = promedio_variables(datos["Temperatura"], L)

print("El promedio es: ", promedio)
```
